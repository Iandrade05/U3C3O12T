```python
def cifra_cesar(texto, chave, modo='cifrar'):
    resultado = ""
    for char in texto:
        if char.isalpha():
            # Define se a letra é maiúscula (A=65) ou minúscula (a=97)
            base = ord('A') if char.isupper() else ord('a')
            # Aplica o deslocamento circular (módulo 26)
            if modo == 'cifrar':
                novo_char = chr((ord(char) - base + chave) % 26 + base)
            else:
                novo_char = chr((ord(char) - base - chave) % 26 + base)
            resultado += novo_char
        else:
            resultado += char
    return resultado

# Exemplo de uso:
mensagem = str(input("Digite a frase:"))
deslocamento = 3

texto_cifrado = cifra_cesar(mensagem, deslocamento, 'cifrar')
texto_decifrado = cifra_cesar(texto_cifrado, deslocamento, 'decifrar')

print(f"Original: {mensagem}")
print(f"Cifrado:  {texto_cifrado}")
print(f"Decifrado: {texto_decifrado}")
```
